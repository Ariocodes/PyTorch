### Matrix multiplication

Two main ways of performing multiplication in neural networks and deep learning:

1. Element-wise multiplication
2. Matrix multiplication (dot product)

There are two main rules that performing matrix multiplication needs to satisfy:
1. The **inner dimensions** must match:
* `(3, 2) @ (3, 2)` won't work
* `(2, 3) @ (3, 2)` will work
* `(3,2) @ (2, 3)` will work
2. The resulting matrix has the shape of the **outer dimension**:
* `(2, 3) @ (3, 2)` -> `(2, 2)`
* `(3, 2) @ (2, 3)` -> `(3, 3)`

```python
torch.matmul(torch.rand(3, 2), torch.rand(3, 2))
```

### One of the most common errors in deep learning: shape errors

```python
# Shapes for matrix multiplication
tensor_A = torch.tensor([[1, 2],
                         [3, 4],
                         [5, 6]])
tensor_B = torch.tensor([[7, 10],
                         [8, 11],
                         [9, 12]])

torch.mm(tensor_A, tensor_B)
```

To fix our tensor shape issues, we can manipulate the shape of one of our tensors using a **transpose**.

A **transpose** switches axes or diensions of a given tensor

it is used as: `myTensor.T`

```python
tensor_B, tensor_B.shape
```

```python
tensor_B.T, tensor_B.T.shape
```

```python
# The matrix multipliatin opreation works when tensor_B is transposed
print(f"Original shapes:\ntensor_A = {tensor_A.shape},\ntensor_B = {tensor_B.shape}")
print(f"\nNew shapes:\ntensor_A = {tensor_A.shape} (same shape as the above),\ntensor_B.T = {tensor_B.T.shape}")
print(f"\nMultiplying:\n{tensor_A.shape} @ {tensor_B.T.shape} <- inner dimensions must match")
print("Output:\n")
output = torch.mm(tensor_A, tensor_B.T)
print(output)
print(f"\nOutput shape: {output.shape}")
```

## Reshaping, stacking, squeezing, and unsqeezing tensors

* Reshaping - reshapes an input tensor to a defined shape
* View - return a view of an input tensor of certain shape but keep the same memory as the original tensor
* Stacking - combine multiple tensors on top of each other (vstack) or side by side (hstack)
* Squeeze - remove all `1` dimensions from a tensor
* Unsqueeze - add a `1` dimension to a target tensor
* Permute - return a view of the input with dimensions permuted (swapped) in a certain way

```python
# Create a tensor
import torch
x = torch.arange(1., 13.)
x, x.shape
```

```python
# Add an extra dimension
# The number of rows times the number of columns must be equal to the total number of elements we have for the reshape method to work.
x_reshaped = x.reshape(1, 12)
x_reshaped, x_reshaped.shape, x_reshaped.ndim
```

```python
x_reshaped2 = x.reshape(12, 1)
x_reshaped2, x_reshaped2.shape, x_reshaped2.ndim
```

```python
x_reshaped3 = x.reshape(3, 4)
x_reshaped3, x_reshaped3.shape, x_reshaped.ndim
```

```python
# Changing the "view"
z = x.view(1, 12)
z, z.shape
```

```python
# Changing z changes x because a view of a tensor shares the same memory location as the original input
z[:, 0] = 5
z, x
```

```python
# Stack tensors on top of each other
x_stacked = torch.stack([x, x, x, x], dim=0) # dim=0 is horizontal, dim=1 is vertical, and so on (depending on how many dimensions our tensor has)
x_stacked
```

```python
x_stacked2 = torch.stack([x, x, x, x], dim=1)
x_stacked2
```

```python
# torch.squeeze() - removes all single dimensions from a target tensor
x = torch.zeros(2, 1, 2, 1, 2)
x.size()
```

```python
y = torch.squeeze(x)
y.size()
```

```python
y = torch.squeeze(x, 0)
y.size()
```

```python
y = torch.squeeze(x, 1)
y.size()
```

```python
y = torch.squeeze(x, (1, 2, 3))
y.size()
```

```python
print(f"Previous tensor: {x_reshaped}")
print(f"Previous shape: {x_reshaped.shape}")

x_squeezed = x_reshaped.squeeze()
print(f"\nNew tensor: {x_squeezed}")
print(f"New shape: {x_squeezed.shape}")
```

```python
# torch.unsqueeze() - adds a asingle dimension to a target tensor at a specific dim
print(f"Previous target: {x_squeezed}")
print(f"Previous shape: {x_squeezed.shape}")

# Add an extra dimension with unsqueeze
x_unsqueezed = x_squeezed.unsqueeze(dim=0)
print(f"\nNew tensor: {x_unsqueezed}")
print(f"New shape: {x_unsqueezed.shape}")
```

```python
# torch.permute - rearranges the dimensions of a target tensor in a specified order
z = torch.randn(2, 3, 5)
print(f"Previous size: {z.size()}")
z_permuted = torch.permute(z, (2, 0, 1)).size() # rearranges the dimensions based on index
print(f"\nNew size: {z_permuted}")
```

```python
x_original = torch.rand(size=(224, 224, 3)) # [height, width, color_channels]

# permute the original tensor to rearrange the axis (or dim) order
x_permuted = x_original.permute(2, 0, 1)
# shifts:
# 2 -> 0
# 0 -> 1
# 1 -> 2

print(f"Previous shape: {x_original.shape}") # [height, width, color_channel]
print(f"New shape:      {x_permuted.shape}") # [color_channel, height, width]
```

```python
x_original[:, 0, 0], x_permuted[:, 0, 0]
```

```python
x_original[0, 0, 0] = 2
x_permuted[0, 0, 0]
```

## PyTorch tensors & NumPy

NumPy is a popular Python numerical computing library.

And because of this, PyTorch has functionality to interact with it.
* Parsing data from NumPy to PyTorch tensor -> `torch.from_numpy(ndarray)`
* Parsing data from PyTorch tensor to Numpy -> `torch.Tensor.numpy()`

```python
import torch
import numpy as np

array = np.arange(1.0, 8.0)
tensor = torch.from_numpy(array)
# The default NumPy datatype is float64, and pytorch reflects numpy's default unless specified otherwise
array, tensor
```

```python
# PyTorch by default creates tensors in float32
torch.arange(1.0, 8.0).dtype
```

```python
# We can force the datatype for pytorch like the following
tensor = torch.from_numpy(array).type(torch.float32)
tensor.dtype
```

```python
# Change the value of array, what will happen to the tensor?
array = array + 1
array, tensor
# tensor is a new memory location so changing the numpy array won't affect the tensor
```

```python
# Tensor to NumPy array
tensor = torch.ones(7)
numpy_tensor = tensor.numpy()
tensor, numpy_tensor
```

```python
# the tensor's datatype is taken by numpy array.
numpy_tensor.dtype
```

```python
# Change the tensor, what happens to numpy_tensor?
tensor = tensor + 1
tensor, numpy_tensor
```

## Reproducibility (trying to take random out of random)
In short how a neural network learns:

`start with random numbers`
->
`tensor operations`
->
`update random numbers to try and make them better representations of the data`
->
`again` -> `again` -> `again...`

To reduce the randomness in neural networks and PyTorch comes the concept of a **random seed**.

Essentially what the random seed does is "flavor" the randomness.

```python
import torch

# Create two random tensors
random_tensor_A = torch.rand(3, 4)
random_tensor_B = torch.rand(3, 4)

print(random_tensor_A)
print(random_tensor_B)
print(random_tensor_A == random_tensor_B)
```

```python
# Let's make some random but reproducible tensors
import torch

# Set the random seed
RANDOM_SEED = 42
torch.manual_seed(RANDOM_SEED)

random_tensor_C = torch.rand(3, 4)
random_tensor_D = torch.rand(3, 4)

print(random_tensor_C)
print(random_tensor_D)
print(random_tensor_C == random_tensor_D)
```

```python
# TRY SETTING MANUAL SEED EACH TIME
torch.manual_seed(RANDOM_SEED) # should use this line every time
random_tensor_E = torch.rand(3, 4)

torch.manual_seed(RANDOM_SEED)
random_tensor_F = torch.rand(3, 4)

print(random_tensor_E)
print(random_tensor_F)
print(random_tensor_E == random_tensor_F)
```
