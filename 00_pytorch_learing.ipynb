### Random tensors
random tensors are important because the way many neural networks learn is that they start with tensors full of random numbers and then adjust those random numbers to better represent the data.


`start with random numbers -> look at data -> update random numbers -> look at data -> update random numbers
`

```python
# create a random tensor of shape (3, 4)
random_tensor = torch.rand(3, 4)
random_tensor
```

```python
random_tensor.ndim
```

```python
# Create a random tensor with similar shape to an image tensor
random_image_size_tensor = torch.rand(size=(3, 224, 224)) # color channel (R, G, B), height, width
random_image_size_tensor.shape, random_image_size_tensor.ndim
```

### Tensor datatypes

*   Sinlge percision: 32-bit float (default)
*   half percision: 16-bit float
*   doule percision: 64-bit float

**Lower percision is faster but less precise and higher percision is vice versa.**


**Note:** Tensor datatypes is one fo the 3 big errors you'll run into with PyTorch & deep learning
1. Tensors not right datatype
2. Tensors not right shape
3. Tensors nto on the right device

```python
# Float 32 tensor
float_32_tensor = torch.tensor([3.0, 6.0, 9.0],
                               dtype=None, # what datatype is the tensor (float32 or float 16)
                               device=None, # what device the tensor is on. "cpu" or "cuda" (cuda: using nvidia gpu)
                               requires_grad=False # whether or not to track gradients with this tensors operations
                               )
float_32_tensor
```

```python
float_32_tensor.dtype
```

```python
# Float 16 tensor
float_16_tensor = torch.tensor([3.0, 6.0, 9.0], dtype=torch.float16)
# OR
float_16_tensor = float_32_tensor.type(torch.float16)
# OR
float_16_tensor = float_32_tensor.type(torch.half)
float_16_tensor
```

```python
float_16_tensor.dtype
```

```python
float_16_tensor * float_32_tensor
```

```python
int_32_tensor = torch.tensor([3, 6, 9], dtype=torch.int32)
int_32_tensor
```

```python
float_32_tensor * int_32_tensor
```

### Getting information from tensors (tensor attributes)
1. Tensors not right datatype - to get datatype from a tesor, can use `tensor.dtype`
2. Tensors not right shape - to get shape from a tensor, can use `tensor.shape`
3. Tensors nto on the right device - to get device from a tensor, can use `tensor.device`

```python
some_tensor = torch.rand(size=(3, 4))
some_tensor
```

`some_tensor.shape()` returns the exact same response as `some_tensor.shape`

(one is a method and the other is an attribute)

```python
# Find out details about some tensor
print(some_tensor, end="\n\n")
print(f"Datatype of tensor: {some_tensor.dtype}")
print(f"Shape of tensor: {some_tensor.shape}")
print(f"Device tensor is on: {some_tensor.device}")
```
